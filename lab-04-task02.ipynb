```python
import pandas as pd

DATA_PATH = "/kaggle/input/datasets/nooresahar08/iris-dataset/Iris.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
```

```python
feature_cols = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

X = df[feature_cols].values
```

```python
import matplotlib.pyplot as plt

# Make sure your DataFrame 'df' is already loaded before running this
plt.scatter(df["PetalLengthCm"], df["PetalWidthCm"], s=20, alpha=0.7)
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Petal Length vs Petal Width")  # Added a title for better visualization
plt.show()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(X)
```

```python
df["Cluster"] = kmeans.predict(X)

print(df["Cluster"].value_counts())
```

```python
comparison = pd.crosstab(df["Cluster"], df["Species"])

print(comparison)
```
